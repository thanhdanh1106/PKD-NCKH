## 3. Task A – Classification (avg AUROC / avg F1)

```python
import subprocess, sys, os

CLS_DIR = os.path.join(os.getcwd(), "downstream_task", "classification")

proc_cls = subprocess.Popen(
    [
        sys.executable, "cls.py",
        "--max_epochs",         "20",
        "--task_type",          "multilabel",
        "--openi",              "True",
        "--data_path",          "../../data/dataset/openi",
        "--Train_dset_name",    "Train.jsonl",
        "--Valid_dset_name",    "Test.jsonl",
        "--loaddir",            "../../saved_models/1",
        "--save_name",          "openi_cls",
        "--batch_sz",           "20",
        "--num_image_embeds",   "180",
        "--n_workers",          "0",
    ],
    cwd=CLS_DIR,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1,
)

cls_output_lines = []
for line in proc_cls.stdout:
    line = line.rstrip("\n")
    print(line)
    cls_output_lines.append(line)

proc_cls.wait()
if proc_cls.returncode != 0:
    print(f"\n[ERROR] cls.py exited with code {proc_cls.returncode}")
```

```python
import os, glob, csv, pandas as pd, numpy as np
from scipy import stats

# cls.py saves CSV to downstream_task/classification/output/.../openi_cls.csv
csv_files = sorted(glob.glob(
    "downstream_task/classification/output/**/*.csv", recursive=True
))
assert csv_files, "No CSV result file found – check that training finished."
latest_csv = csv_files[-1]
print("Reading:", latest_csv)

df_cls = pd.read_csv(latest_csv)
print(df_cls.to_string(index=False))
```

```python
import pandas as pd
import numpy as np
from scipy import stats

# ------------------------------------------------------------------
# Build the Classification result table  (ALL / CHEST split)
# ALL  = every class in the OpenI dataset
# CHEST = only thoracic / lung / pleural pathology classes
# ------------------------------------------------------------------
CHEST_CLASSES = {
    "Cardiomegaly", "Pleural Effusion", "Pneumonia", "Pneumothorax",
    "Atelectasis", "Consolidation", "Edema", "Emphysema",
    "Fibrosis", "Pleural Thickening", "Nodule", "Mass", "Hernia"
}

# Giả định df_cls là DataFrame của bạn
row = df_cls.iloc[-1]          # last row = best / final epoch

exclude_cols = ["micro_auc", "macro_auc", "avg_auroc", "micro_f1", "macro_f1",
                "avg_f1", "avg_precision", "avg_recall", "p_value_auroc", "p_value_f1"]

all_auroc_cols = [c for c in df_cls.columns if c not in exclude_cols]
chest_auroc_cols = [c for c in all_auroc_cols if c in CHEST_CLASSES]

# Tính toán giá trị trung bình
avg_auroc_all   = row[all_auroc_cols].astype(float).mean()
avg_auroc_chest = row[chest_auroc_cols].astype(float).mean() if chest_auroc_cols else float("nan")
avg_f1          = float(row.get("avg_f1", np.nan))

# p-value vs. 0.5 baseline (one-sample t-test)
auroc_vals = row[all_auroc_cols].astype(float).values

# Tránh lỗi nếu auroc_vals có ít hơn 2 phần tử
if len(auroc_vals) > 1:
    p_auroc = stats.ttest_1samp(auroc_vals, 0.5).pvalue
else:
    p_auroc = float("nan")

p_auroc_str = f"{p_auroc:.4e}" if pd.notna(p_auroc) else "NaN"

# ------------------------------------------------------------------
# In bảng kết quả tự động căn lề thẳng hàng
# ------------------------------------------------------------------
print("\n╔══════════════════════════════════════════════════════════╗")
print("║           Classification Results – OpenI Dataset         ║")
print("╠══════════════════════════╦══════════════╦════════════════╣")
print("║          Metric          ║      ALL     ║      CHEST     ║")
print("╠══════════════════════════╬══════════════╬════════════════╣")
print(f"║ {'avg AUROC':<24} ║ {avg_auroc_all:^12.4f} ║ {avg_auroc_chest:^14.4f} ║")
print(f"║ {'avg F1':<24} ║ {avg_f1:^12.4f} ║ {'-':^14} ║")
print(f"║ {'p-value (avg AUROC)':<24} ║ {p_auroc_str:^12} ║ {'-':^14} ║")
print(f"║ {'p-value (avg F1)':<24} ║ {'N/A (1 run)':^12} ║ {'-':^14} ║")
print("╚══════════════════════════╩══════════════╩════════════════╝")

# Tùy chọn: In bảng AUROC cho từng class
print("\nPer-class AUROC:")
per_class = pd.DataFrame({
    "Class": all_auroc_cols,
    "AUROC": row[all_auroc_cols].astype(float).values
})
per_class = per_class.sort_values("AUROC", ascending=False).reset_index(drop=True)
per_class.index += 1
print(per_class.to_string())
```

## 4. Task B – Image–Text Retrieval (MRR / H@5 / R@5 / P@)

```python
RET_DIR = os.path.join(os.getcwd(), "downstream_task", "retrieval")

proc_ret = subprocess.Popen(
    [
        sys.executable, "retrieval.py",
        "--epochs",                             "20",
        # OpenI T2I direction (default: t2i=True, i2t=False)
        "--eval_len_size",                      "354",  # OpenI test size
        "--train_dataset",                      "../../data/dataset/openi/Train.jsonl",
        "--label_conditioned_valid_dataset",    "../../data/dataset/openi/T2I_Label_Test.jsonl",
        "--label_conditioned_test_dataset",     "../../data/dataset/openi/T2I_Label_Test.jsonl",
        "--eval_during_training",               "True",
        "--do_train",                           "True",
        "--num_image_embeds",                   "180",
        "--seq_len",                            "253",
        "--img_channel",                        "3",
        "--bert_model",                         "bert-base-scratch",
        "--batch_size",                         "10",
        "--num_workers",                        "0",
        "--seed",                               "123",
    ],
    cwd=RET_DIR,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1,
)

ret_output_lines = []
for line in proc_ret.stdout:
    line = line.rstrip("\n")
    print(line)
    ret_output_lines.append(line)

proc_ret.wait()
result_ret_stdout = "\n".join(ret_output_lines)
if proc_ret.returncode != 0:
    print(f"\n[ERROR] retrieval.py exited with code {proc_ret.returncode}")
```

```python
import re

stdout_ret = result_ret_stdout   # captured from the streaming cell above

def extract(pattern, text, default="N/A"):
    m = re.search(pattern, text)
    return m.group(1) if m else default

mrr    = extract(r"mrr_score:\s*([\d.]+)", stdout_ret)
r1     = extract(r"['\"]R@1['\"]\s*:\s*([\d.]+)", stdout_ret)
r5     = extract(r"['\"]R@5['\"]\s*:\s*([\d.]+)", stdout_ret)
r10    = extract(r"['\"]R@10['\"]\s*:\s*([\d.]+)", stdout_ret)
p5     = extract(r"['\"]P@1['\"]\s*:\s*([\d.]+)", stdout_ret)
p_mrr  = extract(r"p_mrr['\"]?\s*:\s*([\d.e+-]+)", stdout_ret)
p_h5   = extract(r"p_h5['\"]?\s*:\s*([\d.e+-]+)", stdout_ret)

h5 = r5

print("\n╔══════════════════════════════════════════════════════════╗")
print("║         Retrieval Results – OpenI Dataset                ║")
print("╠══════════════════════╦═══════════════════════════════════╣")
print(f"║  MRR                 ║  {mrr:<33}║")
print(f"║  H@5 (Hits@5)        ║  {h5:<33}║")
print(f"║  R@5 (Recall@5)      ║  {r5:<33}║")
print(f"║  P@1 (Precision@1)   ║  {p5:<33}║")
print(f"║  R@1                 ║  {r1:<33}║")
print(f"║  R@10                ║  {r10:<33}║")
print("╠══════════════════════╬═══════════════════════════════════╣")
print(f"║  p-value (MRR)       ║  {p_mrr:<33}║")
print(f"║  p-value (H@5)       ║  {p_h5:<33}║")
print("╚══════════════════════╩═══════════════════════════════════╝")
```

## 5. Task C – Report Generation & VQA
### (Perplexity ↓ / Accuracy ↑ / Precision ↑ / Recall ↑ / F1 ↑ / BLEU-4 ↑)

```python
RG_DIR = os.path.join(os.getcwd(), "downstream_task", "report_generation_and_vqa")

# finetune.py requires torchrun env vars for single-GPU
rg_env = os.environ.copy()
rg_env["LOCAL_RANK"] = "0"
rg_env["RANK"] = "0"
rg_env["WORLD_SIZE"] = "1"

proc_rg = subprocess.Popen(
    [
        sys.executable, "finetune.py",
        "--tasks",              "report_generation",
        "--generation_dataset", "openi",
        "--num_train_epochs",   "20",
        "--image_root",         "../../data/dataset/open_i/512_3ch",
        "--train_batch_size",   "4",
        "--num_workers",        "0",
        "--seed",               "123",
        "--output_dir",         "output/report_gen/",
    ],
    cwd=RG_DIR,
    env=rg_env,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1,
)

for line in proc_rg.stdout:
    print(line.rstrip("\n"))

proc_rg.wait()
if proc_rg.returncode != 0:
    print(f"\n[ERROR] finetune.py (generation) exited with code {proc_rg.returncode}")
```

```python
proc_gen = subprocess.Popen(
    [
        sys.executable, "generation_decode.py",
        "--model_recover_path", "output/report_gen/model.19.bin",
        "--tasks",              "report_generation",
        "--image_root",         "../../data/dataset/open_i/512_3ch",
        "--generation_dataset", "openi",
        "--num_workers",        "0",
    ],
    cwd=RG_DIR,
    env=rg_env,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1,
)

for line in proc_gen.stdout:
    print(line.rstrip("\n"))

proc_gen.wait()
if proc_gen.returncode != 0:
    print(f"\n[ERROR] generation_decode.py exited with code {proc_gen.returncode}")
```

```python
import pickle, glob

# generation_decode.py saves pickle to its CWD (downstream_task/report_generation_and_vqa/)
gen_pickles = sorted(
    glob.glob("downstream_task/report_generation_and_vqa/*gen*.pickle") +
    glob.glob("downstream_task/report_generation_and_vqa/all_random*_gen.pickle")
)
assert gen_pickles, "No generation pickle found – check that generation_decode.py ran."
with open(gen_pickles[-1], "rb") as f:
    rd_gen = pickle.load(f)

def _mean(lst): return float(np.mean(lst)) if lst else float("nan")
def _std(lst):  return float(np.std(lst))  if len(lst) > 1 else 0.0

print("\n╔══════════════════════════════════════════════════════════════╗")
print("║   Report Generation Results (TABLE VIII style) – OpenI      ║")
print("╠═══════════════════╦═══════════════╦══════════════════════════╣")
print("║  Metric           ║  Value (mean) ║  Std                     ║")
print("╠═══════════════════╬═══════════════╬══════════════════════════╣")
print(f"║  Perplexity (↓)   ║  {_mean(rd_gen['ppl']):<13.4f}║  {_std(rd_gen['ppl']):<24.4f}║")
print(f"║  Accuracy   (↑)   ║  {_mean(rd_gen['accuracy']):<13.4f}║  {_std(rd_gen['accuracy']):<24.4f}║")
print(f"║  Precision  (↑)   ║  {_mean(rd_gen['precision']):<13.4f}║  {_std(rd_gen['precision']):<24.4f}║")
print(f"║  Recall     (↑)   ║  {_mean(rd_gen['recall']):<13.4f}║  {_std(rd_gen['recall']):<24.4f}║")
print(f"║  F1 Score   (↑)   ║  {_mean(rd_gen['f1']):<13.4f}║  {_std(rd_gen['f1']):<24.4f}║")
print(f"║  BLEU-1     (↑)   ║  {_mean(rd_gen['bleu1']):<13.4f}║  {_std(rd_gen['bleu1']):<24.4f}║")
print(f"║  BLEU-2     (↑)   ║  {_mean(rd_gen['bleu2']):<13.4f}║  {_std(rd_gen['bleu2']):<24.4f}║")
print(f"║  BLEU-3     (↑)   ║  {_mean(rd_gen['bleu3']):<13.4f}║  {_std(rd_gen['bleu3']):<24.4f}║")
print(f"║  BLEU-4     (↑)   ║  {_mean(rd_gen['bleu4']):<13.4f}║  {_std(rd_gen['bleu4']):<24.4f}║")
print("╚═══════════════════╩═══════════════╩══════════════════════════╝")
```

## 6. Task C (continued) – VQA-RAD
### TABLE VIII: O.E. = Open-Ended · C.E. = Closed-Ended

```python
proc_vqa = subprocess.Popen(
    [
        sys.executable, "finetune.py",
        "--tasks",              "vqa",
        "--vqa_rad",            "all",
        "--num_train_epochs",   "20",
        "--image_root",         "../../data/dataset/open_i/512_3ch",
        "--train_batch_size",   "4",
        "--num_workers",        "0",
        "--seed",               "123",
        "--output_dir",         "output/vqa/",
    ],
    cwd=RG_DIR,
    env=rg_env,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1,
)

for line in proc_vqa.stdout:
    print(line.rstrip("\n"))

proc_vqa.wait()
if proc_vqa.returncode != 0:
    print(f"\n[ERROR] finetune.py (vqa) exited with code {proc_vqa.returncode}")
```

```python
# finetune.py (vqa) saves pickle to its CWD (downstream_task/report_generation_and_vqa/)
vqa_pickles = sorted(
    glob.glob("downstream_task/report_generation_and_vqa/vqa_*.pickle") +
    glob.glob("downstream_task/report_generation_and_vqa/final_vqa_*.pickle")
)
assert vqa_pickles, "No VQA pickle found – check that VQA eval ran."
with open(vqa_pickles[-1], "rb") as f:
    rd_vqa = pickle.load(f)

oe  = rd_vqa.get("open_acc",   [])
ce  = rd_vqa.get("closed_acc", [])
tot = rd_vqa.get("total_acc",  [])

def _pval(vals):
    if len(vals) < 2: return float("nan")
    from scipy import stats as _st
    _, p = _st.ttest_1samp(vals, np.mean(vals))
    return float(p)

print("\n╔══════════════════════════════════════════════════════════════════════╗")
print("║  TABLE VIII – VQA-RAD Results (* all visual features, OpenI pretrain)║")
print("╠══════════════════╦════════════════╦════════════════╦═════════════════╣")
print("║  Metric          ║  O.E. (Open)   ║  C.E.(Closed)  ║  Total Acc      ║")
print("╠══════════════════╬════════════════╬════════════════╬═════════════════╣")
print(f"║  Accuracy (↑)    ║  {_mean(oe):.4f}         ║  {_mean(ce):.4f}         ║  {_mean(tot):.4f}          ║")
print("╠══════════════════╬════════════════╬════════════════╬═════════════════╣")
print(f"║  p-value O.E.    ║  {_pval(oe):.4e}   ║                ║                 ║")
print(f"║  p-value C.E.    ║                ║  {_pval(ce):.4e}   ║                 ║")
print("╚══════════════════╩════════════════╩════════════════╩═════════════════╝")
```

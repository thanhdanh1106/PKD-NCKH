```python
import os, textwrap

REPO_ROOT = os.getcwd()

# ── 1. Restore downstream_task/classification/data/dataset.py ──────────────
dataset_file = os.path.join(REPO_ROOT, "downstream_task", "classification", "data", "dataset.py")

correct_dataset = textwrap.dedent("""\
    import json
    import numpy as np
    import os
    from PIL import Image

    import torch
    from torch.utils.data import Dataset

    from utils.utils import truncate_seq_pair, numpy_seed


    class JsonlDataset(Dataset):
        def __init__(self, data_path, tokenizer, transforms, vocab, args):
            self.data = [json.loads(l) for l in open(data_path)]
            self.data_dir = os.path.dirname(data_path)
            self.tokenizer = tokenizer
            self.args = args
            self.vocab = vocab
            self.n_classes = len(args.labels)
            self.text_start_token = ["[SEP]"]

            with numpy_seed(0):
                for row in self.data:
                    if np.random.random() < args.drop_img_percent:
                        row["img"] = None

            self.max_seq_len = args.max_seq_len
            self.max_seq_len -= args.num_image_embeds
            self.transforms = transforms

        def __len__(self):
            return len(self.data)

        def __getitem__(self, index):
            sentence = (
                self.text_start_token
                + self.tokenizer(self.data[index]["text"])[: (self.max_seq_len - 1)]
                + self.text_start_token
            )
            segment = torch.zeros(len(sentence))
            sentence = torch.LongTensor(
                [self.vocab.stoi[w] if w in self.vocab.stoi else self.vocab.stoi["[UNK]"]
                 for w in sentence]
            )
            if self.args.task_type == "multilabel":
                label = torch.zeros(self.n_classes)
                if self.data[index]["label"] == '':
                    self.data[index]["label"] = "'Others'"
                label[
                    [self.args.labels.index(tgt)
                     for tgt in self.data[index]["label"].split(', ')]
                ] = 1
            else:
                label = torch.tensor(0)

            image = None
            if self.data[index]["img"]:
                img_field = self.data[index]["img"]
                if os.path.isabs(img_field):
                    img_path = img_field
                elif getattr(self.args, 'img_root_dir', None):
                    img_path = os.path.join(self.args.img_root_dir, img_field)
                else:
                    img_path = os.path.join(self.data_dir, img_field)
                image = Image.open(img_path)
            else:
                image = Image.fromarray(128 * np.ones((256, 256, 3), dtype=np.uint8))
            image = self.transforms(image)

            segment = segment[1:]
            sentence = sentence[1:]
            segment += 1

            return sentence, segment, image, label
    """)

with open(dataset_file, "w", encoding="utf-8") as f:
    f.write(correct_dataset)
print("downstream_task/classification/data/dataset.py restored.")

# ── 2. Restore downstream_task/classification/models/image.py ──────────────
cls_image_file = os.path.join(REPO_ROOT, "downstream_task", "classification", "models", "image.py")

correct_cls_image = textwrap.dedent("""\
    import torch
    import torch.nn as nn
    import torchvision
    from torchvision.models import ResNet50_Weights

    class ImageEncoder(nn.Module):
        def __init__(self, args):
            super(ImageEncoder, self).__init__()
            self.args = args
            model = torchvision.models.resnet50(weights=ResNet50_Weights.IMAGENET1K_V1)
            modules = list(model.children())[:-2]
            self.model = nn.Sequential(*modules)
            # AdaptiveAvgPool1d works with any num_image_embeds (e.g. 180)
            self.pool = nn.AdaptiveAvgPool1d(args.num_image_embeds)

        def forward(self, x):
            out = self.model(x)          # B x 2048 x H x W
            B, C, H, W = out.shape
            out = out.view(B, C, H * W)  # B x 2048 x (H*W)
            out = self.pool(out)          # B x 2048 x num_image_embeds
            return out.transpose(1, 2).contiguous()  # B x num_image_embeds x 2048


    class ImageClf(nn.Module):
        def __init__(self, args):
            super(ImageClf, self).__init__()
            self.args = args
            self.img_encoder = ImageEncoder(args)
            self.clf = nn.Linear(args.img_hidden_sz * args.num_image_embeds, args.n_classes)

        def forward(self, x):
            x = self.img_encoder(x)
            x = torch.flatten(x, start_dim=1)
            out = self.clf(x)
            return out
    """)

with open(cls_image_file, "w", encoding="utf-8") as f:
    f.write(correct_cls_image)
print("downstream_task/classification/models/image.py restored.")

# ── 3. Patch downstream_task/classification/cls.py (macro AUROC nanmean) ────
cls_file = os.path.join(REPO_ROOT, "downstream_task", "classification", "cls.py")
with open(cls_file, "r", encoding="utf-8") as f:
    cls_src = f.read()

old_snippet = (
    "            except ValueError:\n"
    "                outAUROC.append(0)\n"
    "                pass\n"
    "        for i in range(0, len(outAUROC)):\n"
    "            assert args.n_classes == len(outAUROC)\n"
    "            classACC[args.labels[i]] = outAUROC[i]\n"
    "\n"
    "        metrics[\"micro_roc_auc\"] = roc_auc_score(tgts, preds, average=\"micro\")\n"
    "        metrics[\"macro_roc_auc\"] = roc_auc_score(tgts, preds, average=\"macro\")"
)
new_snippet = (
    "            except ValueError:\n"
    "                outAUROC.append(float('nan'))\n"
    "        for i in range(0, len(outAUROC)):\n"
    "            assert args.n_classes == len(outAUROC)\n"
    "            classACC[args.labels[i]] = outAUROC[i]\n"
    "\n"
    "        metrics[\"micro_roc_auc\"] = roc_auc_score(tgts, preds, average=\"micro\")\n"
    "        # nanmean: bỏ qua class không có positive sample trong test set\n"
    "        metrics[\"macro_roc_auc\"] = float(np.nanmean(outAUROC))"
)
if old_snippet in cls_src:
    cls_src = cls_src.replace(old_snippet, new_snippet)
    with open(cls_file, "w", encoding="utf-8") as f:
        f.write(cls_src)
    print("cls.py patched (macro AUROC nanmean fix).")
else:
    print("cls.py already patched or pattern not found – skipping.")

# ── 4. Fix models/train_origin.py (AdamW import) ─────────────────────────────
train_file = os.path.join(REPO_ROOT, "models", "train_origin.py")
with open(train_file, "r", encoding="utf-8") as f:
    src = f.read()
if "from transformers.optimization import AdamW" in src:
    src = src.replace("from transformers.optimization import AdamW", "from torch.optim import AdamW")
    with open(train_file, "w", encoding="utf-8") as f:
        f.write(src)
    print("train_origin.py patched (AdamW import fix).")
else:
    print("train_origin.py already patched – skipping.")

# ── 5. Fix models/MedViLL_origin.py (attn_implementation, dtype, output_attentions) ─
medvill_file = os.path.join(REPO_ROOT, "models", "MedViLL_origin.py")
with open(medvill_file, "r", encoding="utf-8") as f:
    src = f.read()

patches = [
    # Ensure attn_implementation is set (idempotent)
    (
        "        bert = BertModel(model_config)\n\n        self.txt_embeddings",
        "        model_config.attn_implementation = 'eager'\n        bert = BertModel(model_config)\n\n        self.txt_embeddings",
    ),
    # Fix dtype cast in get_extended_attn_mask
    (
        "        extended_attn_mask = extended_attn_mask.to(dtype=torch.float16)\n"
        "        extended_attn_mask = (1.0 - extended_attn_mask) * - 10000.0",
        "        extended_attn_mask = (1.0 - extended_attn_mask.float()) * -10000.0",
    ),
    # Fix output_attentions and return value
    (
        "output_hidden_states=False, output_attentions=True)\n"
        "            return encoded_layers[0], self.pooler(encoded_layers[0]), encoded_layers[1]",
        "output_hidden_states=False, output_attentions=False)\n"
        "            return encoded_layers[0], self.pooler(encoded_layers[0]), None",
    ),
]
changed = False
for old, new in patches:
    if old in src:
        src = src.replace(old, new)
        changed = True
if changed:
    with open(medvill_file, "w", encoding="utf-8") as f:
        f.write(src)
    print("MedViLL_origin.py patched.")
else:
    print("MedViLL_origin.py already patched – skipping.")

# ── 6. Fix models/image.py (ImageEncoder_cnn – AdaptiveAvgPool1d) ─────────────
pretrain_image_file = os.path.join(REPO_ROOT, "models", "image.py")
with open(pretrain_image_file, "r", encoding="utf-8") as f:
    src = f.read()
old_fwd = (
    "    def forward(self, x):\n"
    "        out = self.model(x)  # 512x512: torch.Size([16, 2048, 16, 16])\n"
    "        out = torch.flatten(out, start_dim=2).transpose(1, 2).contiguous()\n"
    "        \n"
    "        vis_pe = torch.arange(out.size(1), dtype=torch.long).cuda()\n"
    "        vis_pe = vis_pe.unsqueeze(0).expand(out.size(0), out.size(1))\n"
    "\n"
    "        random_sampling = torch.randperm(out.size(1))[:self.configs['num_image_embeds']]\n"
    "        random_sampling, _ = torch.sort(random_sampling)\n"
    "\n"
    "        random_sample = out[:, random_sampling]\n"
    "        random_position = vis_pe[:, random_sampling]\n"
    "        return random_sample, random_position"
)
new_fwd = (
    "    def forward(self, x):\n"
    "        out = self.model(x)              # B x 2048 x H x W\n"
    "        B, C, H, W = out.shape\n"
    "        out = out.view(B, C, H * W)      # B x 2048 x (H*W)\n"
    "        pool = nn.AdaptiveAvgPool1d(self.configs['num_image_embeds'])\n"
    "        out = pool(out).transpose(1, 2).contiguous()  # B x num_image_embeds x 2048\n"
    "\n"
    "        vis_pe = torch.arange(out.size(1), dtype=torch.long, device=x.device)\n"
    "        vis_pe = vis_pe.unsqueeze(0).expand(B, -1)\n"
    "        return out, vis_pe"
)
if old_fwd in src:
    src = src.replace(old_fwd, new_fwd)
    with open(pretrain_image_file, "w", encoding="utf-8") as f:
        f.write(src)
    print("models/image.py patched (AdaptiveAvgPool1d fix).")
else:
    print("models/image.py already patched – skipping.")

print(f"\nAll fixes verified under: {REPO_ROOT}")
```

```python
import pandas as pd
import numpy as np
from scipy import stats

CHEST_CLASSES = {
    "Cardiomegaly", "Pleural Effusion", "Pneumonia", "Pneumothorax",
    "Atelectasis", "Consolidation", "Edema", "Emphysema",
    "Fibrosis", "Pleural Thickening", "Nodule", "Mass", "Hernia"
}

row = df_cls.iloc[-1]  # last row = best / final epoch

# Cột tổng hợp cần loại ra khỏi per-class AUROC
exclude_cols = {"micro_auc", "macro_auc", "micro_f1", "macro_f1"}
all_auroc_cols  = [c for c in df_cls.columns if c not in exclude_cols]
chest_auroc_cols = [c for c in all_auroc_cols if c in CHEST_CLASSES]

# avg AUROC = macro_auc đã lưu trong CSV (mean of per-class AUROC)
avg_auroc_all   = float(row["macro_auc"])
avg_auroc_chest = row[chest_auroc_cols].astype(float).mean() if chest_auroc_cols else float("nan")

# avg F1 = macro_f1 đã lưu trong CSV
avg_f1 = float(row["macro_f1"])

# p-value vs. 0.5 baseline (one-sample t-test trên per-class AUROC)
auroc_vals = row[all_auroc_cols].astype(float).values
p_auroc = stats.ttest_1samp(auroc_vals, 0.5).pvalue if len(auroc_vals) > 1 else float("nan")
p_auroc_str = f"{p_auroc:.4e}" if pd.notna(p_auroc) else "NaN"

print("\n╔══════════════════════════════════════════════════════════╗")
print("║           Classification Results – OpenI Dataset         ║")
print("╠══════════════════════════╦══════════════╦════════════════╣")
print("║          Metric          ║      ALL     ║      CHEST     ║")
print("╠══════════════════════════╬══════════════╬════════════════╣")
print(f"║ {'avg AUROC':<24} ║ {avg_auroc_all:^12.4f} ║ {avg_auroc_chest:^14.4f} ║")
print(f"║ {'avg F1':<24} ║ {avg_f1:^12.4f} ║ {'-':^14} ║")
print(f"║ {'p-value (avg AUROC)':<24} ║ {p_auroc_str:^12} ║ {'-':^14} ║")
print(f"║ {'p-value (avg F1)':<24} ║ {'N/A (1 run)':^12} ║ {'-':^14} ║")
print("╚══════════════════════════╩══════════════╩════════════════╝")

print("\nPer-class AUROC:")
per_class = pd.DataFrame({
    "Class": all_auroc_cols,
    "AUROC": row[all_auroc_cols].astype(float).values
}).sort_values("AUROC", ascending=False).reset_index(drop=True)
per_class.index += 1
print(per_class.to_string())
```

```python
import re

stdout_ret = result_ret_stdout   # captured from the streaming cell above

def extract(pattern, text, default="N/A"):
    m = re.search(pattern, text)
    return m.group(1) if m else default

mrr    = extract(r"mrr_score:\s*([\d.]+)", stdout_ret)
r1     = extract(r"['\"]R@1['\"]\s*:\s*([\d.]+)", stdout_ret)
r5     = extract(r"['\"]R@5['\"]\s*:\s*([\d.]+)", stdout_ret)
r10    = extract(r"['\"]R@10['\"]\s*:\s*([\d.]+)", stdout_ret)
p5     = extract(r"['\"]P@1['\"]\s*:\s*([\d.]+)", stdout_ret)
p_mrr  = extract(r"p_mrr['\"]?\s*:\s*([\d.e+-]+)", stdout_ret)
p_h5   = extract(r"p_h5['\"]?\s*:\s*([\d.e+-]+)", stdout_ret)

h5 = r5

print("\n╔══════════════════════════════════════════════════════════╗")
print("║         Retrieval Results – OpenI Dataset                ║")
print("╠══════════════════════╦═══════════════════════════════════╣")
print(f"║  MRR                 ║  {mrr:<33}║")
print(f"║  H@5 (Hits@5)        ║  {h5:<33}║")
print(f"║  R@5 (Recall@5)      ║  {r5:<33}║")
print(f"║  P@1 (Precision@1)   ║  {p5:<33}║")
print(f"║  R@1                 ║  {r1:<33}║")
print(f"║  R@10                ║  {r10:<33}║")
print("╠══════════════════════╬═══════════════════════════════════╣")
print(f"║  p-value (MRR)       ║  {p_mrr:<33}║")
print(f"║  p-value (H@5)       ║  {p_h5:<33}║")
print("╚══════════════════════╩═══════════════════════════════════╝")
```

```python
# finetune.py (vqa) saves pickle to its CWD (downstream_task/report_generation_and_vqa/)
vqa_pickles = sorted(
    glob.glob("downstream_task/report_generation_and_vqa/vqa_*.pickle") +
    glob.glob("downstream_task/report_generation_and_vqa/final_vqa_*.pickle")
)
assert vqa_pickles, "No VQA pickle found – check that VQA eval ran."
with open(vqa_pickles[-1], "rb") as f:
    rd_vqa = pickle.load(f)

oe  = rd_vqa.get("open_acc",   [])
ce  = rd_vqa.get("closed_acc", [])
tot = rd_vqa.get("total_acc",  [])

def _pval(vals):
    if len(vals) < 2: return float("nan")
    from scipy import stats as _st
    _, p = _st.ttest_1samp(vals, np.mean(vals))
    return float(p)

print("\n╔══════════════════════════════════════════════════════════════════════╗")
print("║  TABLE VIII – VQA-RAD Results (* all visual features, OpenI pretrain)║")
print("╠══════════════════╦════════════════╦════════════════╦═════════════════╣")
print("║  Metric          ║  O.E. (Open)   ║  C.E.(Closed)  ║  Total Acc      ║")
print("╠══════════════════╬════════════════╬════════════════╬═════════════════╣")
print(f"║  Accuracy (↑)    ║  {_mean(oe):.4f}         ║  {_mean(ce):.4f}         ║  {_mean(tot):.4f}          ║")
print("╠══════════════════╬════════════════╬════════════════╬═════════════════╣")
print(f"║  p-value O.E.    ║  {_pval(oe):.4e}   ║                ║                 ║")
print(f"║  p-value C.E.    ║                ║  {_pval(ce):.4e}   ║                 ║")
print("╚══════════════════╩════════════════╩════════════════╩═════════════════╝")
```
